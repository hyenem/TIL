### 학습 데이터 준비
 - df를 X와 Y로 분리하세요.
   - X: 'target', 'X10'을 제외한 모든 열
   - Y: 'target'

```python
X = df.drop(['target', 'X10'], axis=1)
X
```

```python
# Code here
Y = df['target']
Y
```

### 선형 회귀 모델을 이용한 모델링

```python
# 필수 모델 및 전처리기
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split
```

### 데이터 분할
 - dateset을 trainset과 testset으로 분할하세요.
 - 분할 비율은 train 75 : test 25입니다.
 - random_state = 0

```python
# Code here
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.25, random_state=0)
```

### Linear Regression
 - LinearRegression 클래스를 이용하여 학습을 수행하세요.
 - 학습 데이터와 테스트 데이터의 score를 각각 출력하세요.

```python
# Code here
model = LinearRegression().fit(x_train, y_train)
model.score(x_train, y_train), model.score(x_test, y_test)
```

### Ridge Regression
 - Ridge 클래스를 이용하여 학습을 수행하세요.
 - 학습 데이터와 테스트 데이터의 score를 각각 출력하세요.
 - alpha=5

```python
# Code here
model = Ridge(alpha=5).fit(x_train, y_train)
model.score(x_train, y_train), model.score(x_test, y_test)
```

### Lasso Regression
 - Lasso 클래스를 이용하여 학습을 수행하세요.
 - 학습 데이터와 테스트 데이터의 score를 각각 출력하세요.
 - alpha=0.01

```python
# Code here
model = Lasso(alpha=0.01).fit(x_train, y_train)
model.score(x_train, y_train), model.score(x_test, y_test)
```

### 스케일링 및 다항 특성 변환을 적용한 다항 회귀 모델링
 - X 데이터에 Stanadard Scaling을 적용하세요.
 - 변환된 X에 2, 3차 다항특성 변환을 적용한 후 다시 분할하고, Lasso로 모델링하세요.
 - 적절한 alpha를 설정해보세요.
 - 학습 데이터와 테스트 데이터의 score를 각각 출력하세요.

```python
# Code here
from sklearn.preprocessing import StandardScaler, PolynomialFeatures

X_scaled = StandardScaler().fit_transform(X)

X_scaled
```

```python
#다항회귀 2차 적용해서 각 알고리즘에 입력하여 성능 평가
X_sc_poly = PolynomialFeatures(degree=2).fit_transform(X_scaled)
x_train, x_test, y_train, y_test = train_test_split(X_sc_poly, Y, test_size=0.25, random_state=0)

model1 = LinearRegression().fit(x_train, y_train)
model2 = Ridge(alpha=5).fit(x_train, y_train)
model3 = Lasso(alpha=0.01).fit(x_train, y_train)

print(model1.score(x_train, y_train), model1.score(x_test, y_test))
print(model2.score(x_train, y_train), model2.score(x_test, y_test))
print(model3.score(x_train, y_train), model3.score(x_test, y_test))
```

```python
#결과 아래와 같이 시각화 
import warnings
warnings.filterwarnings(action='ignore')

#다항회귀 3차에 Lasso에 alpha값 : 0.01~0.3까지 간격을 0.01 주면서 성능 평가
train_cost=[]
test_cost=[]

X_th_poly = PolynomialFeatures(degree=3).fit_transform(X_scaled)
x_train, x_test, y_train, y_test = train_test_split(X_th_poly, Y, test_size=0.25, random_state=0)

x = np.arange(0.01, 0.31, 0.01)
for a in x:
    model=Lasso(alpha=a).fit(x_train, y_train)
    train_cost.append(model.score(x_train, y_train))
    test_cost.append(model.score(x_test, y_test))
```

```python
#위의 학습한 결과 시각화 
plt. figure(figsize = (8,6))
plt.plot(x, train_cost, label = "train", color = "b", marker = "^")
plt.plot(x, test_cost, label = "test", color = "r" , marker = ".")
plt.legend()
plt.grid(True)
plt.show()
```
