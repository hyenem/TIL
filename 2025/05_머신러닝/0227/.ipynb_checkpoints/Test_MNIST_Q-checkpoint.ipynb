## 여기서부터 코드를 작성/수정하세요
#### - 데이터셋 로드, 전처리(필요시), 모델생성/컴파일, 학습
#### - 아래 샘플코드들은 실제 시험 시 없을수 있습니다

```python
train = pd.read_csv("data/train.csv")
test = pd.read_csv("data/test.csv")
submission = pd.read_csv("data/submission.csv")
```

```python
train.head()
```

```python


x_train = train.iloc[:,1:]
y_train = train.iloc[:,0]
x_test = test

x_train.shape, y_train.shape, x_test.shape
```

```python
from sklearn.linear_model import LogisticRegression

model_logistic = LogisticRegression(max_iter=1000).fit(x_train, y_train)
model_logistic.score(x_train, y_train)
```

### 생성한 정답의 shape 및 타입 확인

```python
# 아래의 코드는 필요에따라 수정해서 사용합니다
pred = model_logistic.predict(x_test)
your_answer = pred
your_answer
```

```python
submission['Label'] = your_answer
```
