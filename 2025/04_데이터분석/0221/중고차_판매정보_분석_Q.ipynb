# 중고차 판매 정보 분석 실습

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
# vehicles.csv 파일을 읽어 df 변수에 저장하세요.
df = pd.read_csv('../data/vehicles.csv')
df
```

```python
# Unnamed: 0 컬럼을 제거하세요.
df.drop('Unnamed: 0', axis=1, inplace=True)
df
```

```python
# 가격 컬럼의 단위를 만원으로 변경하세요.
df['가격']=df['가격']/10000
df
```

```python
# 제조사별 판매 차량 수를 출력하세요.
df['제조사'].value_counts()
```

```python
# 제조사별 판매 차량 수를 파이차트로 시각화하세요.
df['제조사'].value_counts().plot(kind='pie')
```

```python
# 유형별 판매 차량 수를 출력하세요.
df['유형'].value_counts()
```

```python
# 유형별 판매 차량 수를 bar 차트로 시각화하세요.
df['유형'].value_counts().plot(kind='bar')
```

```python
# 판매 가격이 가장 높은 데이터 6개를 출력하세요.
df.sort_values(by='가격', ascending=False).head(6)
```

```python
# 판매 가격이 가장 높은 6개 제품을 '제조사:모델' 대비 가격의 형태로 bar 차트 시각화하세요.
df['제조사:모델']=df['제조사']+':'+df['모델']
df
df.set_index('제조사:모델').sort_values(by='가격', ascending=False).head(6)['가격'].plot(kind='barh')
```

```python
# 평균 판매 가격이 높은 6개 제품을 '제조사:모델' 대비 가격의 형태로 bar 차트 시각화하세요.
df.groupby(['제조사', '모델'])['가격'].mean().sort_values().head(6).plot(kind='barh')
```

```python
# 연식, 주행거리, 가격 간의 상관 계수를 출력하세요.
df[['연식', '주행거리', '가격']].corr()
```

```python
# 연식, 주행거리, 가격 간의 관계를 pairplot을 이용하여 상태 별로 시각화하세요.
sns.pairplot(df[['연식', '주행거리', '가격', '상태']], hue='상태')
plt.show()
```

```python
# 주행거리 열을 히스토그램으로 시각화하세요.
df['주행거리'].value_counts()
df['주행거리'].value_counts().plot(kind='hist', bins=1000)
```

```python
# 주행거리 열을 로그로 변환하여 주행거리_로그 열에 저장하세요. (그대로 실행하세요)
import numpy as np
df['주행거리_로그']=df['주행거리'].apply(np.log)
df
```

```python
# 주행거리 로그 열을 히스토그램으로 시각화하세요.
sns.histplot(df, x='주행거리_로그')
plt.show()
```

```python
# df에 새 열 '변속기_변환'을 추가하세요.
# 값은 아래 조건에 따라 설정하세요.
# 변속기의 값이 기타면 0
# 변속기의 값이 수동이면 1
# 변속기의 값이 자동이면 2
df['변속기_변환']=df['변속기'].apply(lambda x : 1 if x=='수동' else 2 if x=='자동' else 0)
df['변속기_변환']
```

```python
# 연식, 주행거리, 가격, 변속기 간의 관계를 버블 차트(scatter)로 시각화하세요.
# x=연식, y=주행거리_로그, s=가격/30, c=변속기_변환
plt.figure(figsize=(15, 10))
x=df['연식']
y=df['주행거리_로그']

plt.scatter(x, y, s=df['가격']/30, c=df['변속기_변환'])

plt.show()
```

```python
# 주행거리가 100000보다 큰 데이터만 추출한 후 df_100k에 저장하세요.
df_100k = df.loc[df['주행거리']>100000]
df_100k
```

```python
# df_100k의 유형 별 평균 가격을 계산하고 내림차순 정렬하여 출력하세요.
df_100k.groupby('유형')['가격'].mean().sort_values(ascending=False)
```

```python
# 위 데이터를 bar 차트로 시각화하세요.
bar = df_100k.groupby('유형')['가격'].agg(['mean']).sort_values(by='mean',ascending=False)
bar
sns.barplot(bar, x='유형', y='mean')
```
